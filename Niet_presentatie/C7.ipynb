```python
import numpy as np

MAX_NEIGHBORS = 2


def is_pos_feasible(x, pos):
    i, j = pos
    num_neighbors = 0
    for di in [-1, 0, 1]:
        for dj in [-1, 0, 1]:
            if di == 0 and dj == 0:
                continue
            ni, nj = i + di, j + dj

            ni = ni % x.shape[0]
            nj = nj % x.shape[1]

            num_neighbors += 1 if x[ni, nj] == 1 else 0
            if num_neighbors > MAX_NEIGHBORS:
                return False
    return True


def check_state_still_feasible(x):
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            if x[i, j] == 1:
                if not is_pos_feasible(x, (i, j)):
                    return False
    return True


def find_next_states(x):
    next_states = []
    for i in range(x.shape[0]):
        for j in range(x.shape[1]):
            if x[i, j] == -1:
                new_state = x.copy()
                new_state[i, j] = 1
                if check_state_still_feasible(new_state):
                    next_states.append(new_state)
                new_state = x.copy()
                new_state[i, j] = 0
                if check_state_still_feasible(new_state):
                    next_states.append(new_state)
                return next_states
    return []


def count_colored_cells(x):
    return np.sum(x == 1)


def is_final_state(x):
    return np.all(x != -1)


def find_optimum():
    x = np.ones((7, 7)) * -1
    lower_bound = 22  # Proof by Joppe
    options = [(x, 0)]
    x_opt = None
    print("Commencing search...")
    while options:
        state, num_colored = options.pop()
        if is_final_state(state):
            if num_colored >= lower_bound:
                lower_bound = num_colored
                x_opt = state
                print("New best:", lower_bound)
            continue
        for next_state in find_next_states(state):
            next_colored = count_colored_cells(next_state)
            if next_colored + np.sum(next_state == -1) < lower_bound:
                continue
            options.append((next_state, next_colored))
    print("Optimum found:", lower_bound)
    print(x_opt)



find_optimum()
```

```python
cur_answer = 0
cur_answers = []
d = 4
n = 7
import numpy as np

def check(answers, x):
    for a in answers:
        found = False
        for i in range(len(x)):  # Fixed: was "len(x)" should be "range(len(x))"
            if (1 < abs(x[i]-a[i]) < n-1):
                found = True
                break  # Exit the i loop, move to next a
        if not found:
            return False  # No match found for this answer
    return True  # All answers had at least one match

def check_all(answers):
    for i in range(n**d):
        for j in range(d):
            x[j] = (i // n**j) % n
        if check(answers, x):
            return x
    return None
     



counter = 0
while (cur_answer < 110):
    answer = 0
    answers = []
    while (True):
        x = np.zeros(d, dtype=int)
        for i in range(d):
            x[i] = np.random.randint(0, n-1)
        if check(answers, x):
            answers.append(x)
            answer += 1
            counter = 0
        else:
            counter += 1
        if counter > 10000:
            check_res = check_all(answers)
            if check_res is not None:
                answers.append(check_res)
                answer += 1
                counter = 0
            else:
                break
    print("Current answer:", answer)
    if answer > cur_answer:
        cur_answer = answer
        cur_answers = answers.copy()
    print("Best answer so far:", cur_answer)
```
